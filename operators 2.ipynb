# 6.Net salary caluculator

```python
basic_salary=int(input())
house_rent_alowances=int(input())
dearnress_aloowances=int(input())
allwances=int(input())
provident_fund=int(input())
tax=int(input())
other_deductions=int(input())
GS=basic_salary+house_rent_alowances+dearnress_aloowances+allwances
tax_amount= GS * tax / 100
NS=GS-(provident_fund+tax_amount+other_deductions)
print("Net salary caluculator:",NS)
```

# 20.shopping bill caluculator

```python
price = float(input("Enter total shopping amount: "))
discount = float(input("Enter discount percentage: "))
gst = float(input("Enter GST percentage: "))

discount_amount = (price * discount) / 100

price_after_discount = price - discount_amount

gst_amount = (price_after_discount * gst) / 100

final_amount = price_after_discount + gst_amount

print("Discount Amount =", discount_amount)
print("GST Amount =", gst_amount)
print("Final Amount =", final_amount)
```
