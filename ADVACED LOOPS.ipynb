```python
#1.print prime numbers in a given range
start=int(input())
end=int(input())
for n in range(start,end+1):
    if n>1:
        count=0
        for i in range(1,n+1):
            if n%i==0:
                count+=1
        if count==2:
            print(n)
```

```python
#2.print sum of prime numbers in a given range
start=int(input())
end=int(input())
total=0
for n in range(start,end+1):
    if n>1:
        count=0
        for i in range(1,n+1):
            if n%i==0:
                count+=1
        if count==2:
            total+=n
print(total)
```

```python
#3.print prime factors of a number
n=int(input())
i=2
while n>0:
    if n%i==0:
        print(i,end="")
        n=n//i
    else:
        i=i+1
```

```python
#4.print greatest common divisor of two numbers
a=int(input())
b=int(input())
gcd=1
for i in range(1,min(a,b)+1):
    if a%i==0 and b%i==0:
        gcd=i
print("Greatest common divisior of two numbers:",a,b,"is", gcd)
        
        
```

```python
#5.print least common divisior of two numbers
a=int(input())
b=int(input())
lcm=1
for i in range(1,a*b+1):
    if i%a==0 and i%b==0:
        lcm=i
        break
print("Least common Divisior of two numbers:",a,b,"is",lcm)        
```

```python
#6.check whether  a number is a strong number or not
n=int(input())
temp=n
total=0
while temp>0:
    digit=temp%10
    fact=1
    for i in range(1,digit+1):
        fact=fact*i
    total=total+fact
    temp=temp//10
if total==n:
    print("Strong number")
else:
    print("Not a strong number")
```

```python
#7. print a number is perfect square or not
n=int(input())
i=1
while i*i>0:
    if i*i==n:
        print("perfect square")
        break
    i=i+1
else:
    print("Not a perfect square")
    
```

```python
#8.print a number is perfect cube or not
n=int(input())
i=1
while i*i*i>0:
    if i*i*i==n:
        print("perfect cube")
        break
    i=i+1
else:
    print("not a perfect cube")
    
```

```python
#9.check whether a number is palindrome or not
n=int(input())
temp=n
rev=0
while temp>0:
    digit=temp%10
    rev=rev*10+digit
    temp=temp//10
if rev==n:
    print("palindrome number")
else:
    print("not a palindrome number")
    
```

```python
#10.find the largest digit in a number
n=int(input())
largest=0
while n>0:
    digit=n%10
    if digit>largest:
        largest=digit
    n=n//10
print("largest digit in a number is:",largest)    
    
        
```

```python
#11.find the smallest digit in a number
n=int(input())
smallest=9
while n>0:
    digit=n%10
    if digit<smallest:
        smallest=digit
    n=n//10
print("smallest digit in a number:",smallest)    
```

```python
#12.Find the frequecy of each digit in a number
n=int(input())
for digit in range(10):
    temp=n
    c=0
    while temp>0:
        if temp%10==digit:
            c+=1
        temp=temp//10
    if c>0:
        print(digit,"=",c)
        
        
```

```python
# 13.convert decimal number to binary number
n=int(input())
binary=""
while n>0:
    r=n%2
    binary=str(r)+binary
    n=n//2
print("Binary:",binary)    
```

```python
# 14.convert binary number to decimal
bin=int(input())
dec=0
power=0
while bin>0:
    digit=bin%10
    dec=dec+digit*(2**power)
    power+=1
    bin=bin//10
print("Decimnal",dec)    
    

```

```python
# 15.find the sum of factorial of a digits
n=int(input())
temp=n
sum=0
while temp>0:
    digit=temp%10
    fact=1
    for i in range(1,digit+1):
        fact=fact*i
    sum+=fact
    temp=temp//10
print(sum)    

```
