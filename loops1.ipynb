 # 15. Check Whether a Number is a Palindrome

```python
n = int(input("Enter a number: "))

original = n
reverse = ""

for digit in str(n):
    reverse = digit + reverse

if str(original) == reverse:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

```python
n = int(input("Enter a number: "))

original = n
reverse = ""

for digit in str(n):
    reverse = digit + reverse

if str(original) == reverse:
    print("Palindrome")
else:
    print("Not a Palindrome")
```


# 16. Check Whether a Number is an Armstrong Number


```python
n = int(input("Enter a number: "))

original = n
digits = len(str(n))
sum = 0

for digit in str(n):
    sum = sum + int(digit) ** digits

if original == sum:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

```python
n = int(input("Enter a number: "))

original = n
digits = len(str(n))
sum = 0

while n > 0:
    digit = n % 10
    sum = sum + digit ** digits
    n = n // 10

if original == sum:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```


# 17. Check Whether a Number is a Perfect Number

```python
n = int(input("Enter a number: "))

sum = 0

for i in range(1, n):
    if n % i == 0:
        sum = sum + i

if sum == n:
    print("Perfect Number")
else:
    print("Not a Perfect Number")
```

```python
n = int(input("Enter a number: "))

sum = 0
i = 1

while i < n:
    if n % i == 0:
        sum = sum + i
    i += 1

if sum == n:
    print("Perfect Number")
else:
    print("Not a Perfect Number")
```

 # 18. Generate the Fibonacci Series

```python
n = int(input("Enter number of terms: "))
a = 0
b = 1
for i in range(n):
    print(a, end=" ")
    c = a + b
    a = b
    b = c
```

```python
n = int(input("Enter number of terms: "))

a = 0
b = 1
i = 1

while i <= n:
    print(a, end=" ")

    c = a + b
    a = b
    b = c

    i += 1
    
```

 # 19. Check Whether a Number is Prime

```python
n = int(input("Enter a number: "))
count = 0
for i in range(1, n + 1):
    if n % i == 0:
        count += 1
if count == 2:
    print("Prime Number")
else:
    print("Not a Prime Number")
```

```python
n = int(input("Enter a number: "))
count = 0
i = 1
while i <= n:
    if n % i == 0:
        count += 1
    i += 1
if count == 2:
    print("Prime Number")
else:
    print("Not a Prime Number")
```
