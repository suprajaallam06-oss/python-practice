# SIMPLE INTEREST CALUCULATOR

```python
#QUESTION:
#Write a Python program to calculate Simple Interest using Principal Amount, Rate of Interest, and Time.
p = float(input("Enter principal amount: "))
r = float(input("Enter rate of interest: "))
t = float(input("Enter time in years: "))

si = (p * r * t) / 100

print("Simple Interest =", si)
```

# compound intrest caluculator

```python

#Write a Python program to calculate Compound Interest using Principal Amount, Rate of Interest, and Time.
p = float(input("Enter principal amount: "))
r = float(input("Enter rate of interest: "))
t = float(input("Enter time in years: "))

amount = p * (1 + r / 100) ** t
ci = amount - p

print("Compound Interest =", ci)
```

# DISCOUNT CALUCULATOR

```python

#Write a Python program to calculate the discount amount and final price of a product.
price = float(input("Enter product price: "))
discount = float(input("Enter discount percentage: "))

discount_amount = (price * discount) / 100
final_price = price - discount_amount

print("Discount Amount =", discount_amount)
print("Final Price =", final_price)
```

# GST CALUCULATOR

```python

#Write a Python program to calculate GST amount and final price of a product.
price = float(input("Enter product price: "))
gst = float(input("Enter GST percentage: "))

gst_amount = (price * gst) / 100
final_price = price + gst_amount

print("GST Amount =", gst_amount)
print("Final Price =", final_price)
```

 # 7. Profit and Loss Calculator

```python

 #Write a Python program to calculate Profit or Loss using Cost Price and Selling Price.
cp = float(input("Enter cost price: "))
sp = float(input("Enter selling price: "))

if sp > cp:
    profit = sp - cp
    print("Profit =", profit)
elif cp > sp:
    loss = cp - sp
    print("Loss =", loss)
else:
    print("No Profit No Loss")
```

 # 9. EMI Calculator

```python
"""Write a Python program to calculate the monthly EMI of a loan using Loan Amount,
Annual Interest Rate, and Loan Period in Months."""
p = float(input("Enter loan amount: "))
r = float(input("Enter annual interest rate: "))
n = int(input("Enter loan period in months: "))

monthly_rate = r / (12 * 100)

emi = (p * monthly_rate * (1 + monthly_rate) ** n) / ((1 + monthly_rate) ** n - 1)
print("Monthly EMI =", emi)

```

 # 11. Distance Between Two Points

```python

 #Write a Python program to calculate the distance between two points using their x and y coordinates.
import math

x1 = float(input("Enter x1: "))
y1 = float(input("Enter y1: "))
x2 = float(input("Enter x2: "))
y2 = float(input("Enter y2: "))

distance = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)

print("Distance =", distance)
```

 # 15. Time Converter

```python

#Write a Python program to convert a given number of Seconds into Hours, Minutes, and Seconds.
total_seconds = int(input("Enter total seconds: "))

hours = total_seconds // 3600
remaining = total_seconds % 3600
minutes = remaining // 60
seconds = remaining % 60

print("Hours =", hours)
print("Minutes =", minutes)
print("Seconds =", seconds)
```

 # 17. Electricity Bill Calculator

```python

# Write a Python program to calculate an electricity bill based on the number of units consumed.
units = float(input("Enter electricity units consumed: "))

if units <= 100:
    bill = units * 1.5
elif units <= 200:
    bill = (100 * 1.5) + ((units - 100) * 2.5)
else:
    bill = (100 * 1.5) + (100 * 2.5) + ((units - 200) * 4)

print("Electricity Bill =", bill)
```

# 20. Shopping Bill Calculator

```python

# Write a Python program to calculate Discount, GST, and Final Amount of a Shopping Bill.
price = float(input("Enter total shopping amount: "))
discount = float(input("Enter discount percentage: "))
gst = float(input("Enter GST percentage: "))

discount_amount = (price * discount) / 100

price_after_discount = price - discount_amount

gst_amount = (price_after_discount * gst) / 100

final_amount = price_after_discount + gst_amount

print("Discount Amount =", discount_amount)
print("GST Amount =", gst_amount)
print("Final Amount =", final_amount)
```
