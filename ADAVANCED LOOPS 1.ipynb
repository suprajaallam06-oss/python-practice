 # 1. Print All Prime Numbers in a Given Range

```python
start = int(input("Enter start: "))
end = int(input("Enter end: "))

for n in range(start, end + 1):
    if n > 1:
        count = 0

        for i in range(2, n):
            if n % i == 0:
                count += 1

        if count == 0:
            print(n)
```

```python
start = int(input("Enter start: "))
end = int(input("Enter end: "))

n = start

while n <= end:
    if n > 1:
        i = 2
        count = 0

        while i < n:
            if n % i == 0:
                count += 1
            i += 1

        if count == 0:
            print(n)

    n += 1
```

 # 2. Find the Sum of Prime Numbers in a Given Range

```python
start = int(input("Enter start: "))
end = int(input("Enter end: "))

sum = 0

for n in range(start, end + 1):
    if n > 1:
        count = 0

        for i in range(2, n):
            if n % i == 0:
                count += 1

        if count == 0:
            sum += n

print("Sum =", sum)
```

```python
start = int(input("Enter start: "))
end = int(input("Enter end: "))
sum = 0
n = start
while n <= end:
    if n > 1:
        i = 2
        count = 0

        while i < n:
            if n % i == 0:
                count += 1
            i += 1
        if count == 0:
            sum += n
    n += 1
print("Sum =", sum)
```

 # 3. Print the Prime Factors of a Number

```python
n = int(input("Enter a number: "))
for i in range(2, n + 1):
    if n % i == 0:
        count = 0
        for j in range(2, i):
            if i % j == 0:
                count += 1
        if count == 0:
            print(i)
```

```python
n = int(input("Enter a number: "))
i = 2
while i <= n:
    if n % i == 0:
        j = 2
        count = 0
        while j < i:
            if i % j == 0:
                count += 1
            j += 1
        if count == 0:
            print(i)
    i += 1
```

 # 4. Find the GCD of Two Numbers

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
gcd = 1
for i in range(1, min(a, b) + 1):
    if a % i == 0 and b % i == 0:
        gcd = i
print("GCD =", gcd)
```

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
i = 1
gcd = 1
while i <= min(a, b):
    if a % i == 0 and b % i == 0:
        gcd = i
    i += 1
print("GCD =", gcd)
```
