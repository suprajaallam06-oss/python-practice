# IDENTIFIERS:

```python
"""
A name in python program is called indentifier.
The name of a variable or function name or class name are called Indentifiers"""
print("while".isidentifier())
print("father name".isidentifier())
print("2mother name".isidentifier())
print("mother_name".isidentifier())
```

# Variables

```python
"""
Variables are like containers for storing the values or adress of a value.
--->Donot start with digit or symbol
--->starting letter should be alphabet or underscore
--->variables are case-sensitive
--->python has no command for declaring a variable
--->when we assign value the varaible is created
--->variable donot need to be declared with the type.we can change the type after they have been set...
--->single value can be assinged to multiple varaibles
--->multiple values can be assinged to multiple variables
"""
a=b=c=12
print(a,b,c)
a,b,c=2,3,4
print(a,b,c)
```
