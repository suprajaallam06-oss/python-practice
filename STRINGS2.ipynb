# 5.PROGRAM TO CHECK GIVEN TWO STRINGS ARE ANAGRAM

```python
s1=input()
s2=input()
if len(s1)!=len(s2):
    print("Not Anagrams")
else:
    for i in s1:
        if s1.count(i)!=s2.count(i):
            print("Not anagrams")
    else:
       print("Anagrams")
```

# 7.Frequency of each character in a string

```python
s=input()
for i in range(len(s)):
    if s[i] in s[:i]:
        continue
    frequency=0
    for j in range(len(s)):
        if s[i]==s[j]:
            frequency+=1
    print(s[i],":",frequency)       
            

                
    
```

# 9.Reverse words in a string

```python
s=input()
words=s.split()
result=""
for i in range(len(words)-1,-1,-1):
    result=result+words[i]+" "
print(result)    
    
```

# 12. Find the longestsubstring without repeating characters

```python
s=input()
longest=""
for i in range(len(s)):
    current=""
    for j in range(i,len(s)):
        if s[j] not in current:
            current+=s[j]
        else:
            break
        if len(current)>len(longest):
            longest=current
print(longest)            
    
```
