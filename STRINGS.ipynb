# 1. Find the length of the string without using length function

```python
s=input()
length=0
for i in s:
    length+=1
print(length)    
```

# 2. Program to reverse a string

```python
s=input()
reverse=""
for i in s:
    reverse=i+reverse
print(reverse)    
```

# 3.program to count the number of  vowels and consonents in a string

```python
s=input()
v=0
c=0
for i in s:
    if i in "aeiouAEIOU":
        v+=1
    else:
        c+=1
print("vowels:",v)
print("consonents:",c)
```

# 4.ab@#$%^&23456789dferh6yh

```python
s=input()
r1=""
r2=""
r3=""
for i in s:
    if i.isalpha():
        r1+=i
    elif i.isdigit():
        r2+=i
    else:
        r3+=i
print(r3,r2,r1)        
            
```

# 5.PROGRAM TO CHECK GIVEN TWO STRINGS ARE ANAGRAM

```python
s1=input()
s2=input()
if len(s1)!=len(s2):
    print("Two strings are not Anagrams")
else:
    for i in s1:
        if s1.count(i)!=s2.count(i):
            print("Two strings are not Anagrams:")
    else:
        print("Anagrams")
```

# 7.Frequency of each character in a string

```python
s=input()
for i in range(len(s)):
    if s[i] in s[:i]:
        continue
    frequency=0
    for j in range(len(s)):
        if s[i]==s[j]:
            frequency+=1
    print(s[i] ,":",frequency)        
```

# 8.Removes duplicates of a string

```python
s=input()
r=""
for i in s:
    if i not in r:
        r+=i
print(r)        
```
